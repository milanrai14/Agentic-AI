```python
from langgraph.store.memory import InMemoryStore
```

```python
#creating a store 
store = InMemoryStore()
```

```python
#creation a namespace 
namespace = ("user", "1") #namespace is used to organize and separate memorires into differet groups in LangGrapH
```

```python
# Creating memory
# Adding memories

store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data": "User prefer dark mode"})
```

```python
#another namespace 
namespace2 = ("user","u2" )

#Add memories 
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

```python
#Retriving memorirss
#store.get(namespace, key)
store.get(namespace, "1")
```

```python
store.get(namespace, "2")
```

#Retriveing all the memories

```python
items = store.search(namespace)

for item in items: 
    print(item.value)
```
