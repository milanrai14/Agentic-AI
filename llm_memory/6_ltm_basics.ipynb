```python
from langgraph.store.memory import InMemoryStore
```

```python
#creating a store 
store = InMemoryStore()
```

```python
#creation a namespace 
namespace = ("user", "1")
```

```python
# Creating memory
# Adding memories

store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data": "User prefer dark mode"})
```
