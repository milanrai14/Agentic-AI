```python
from langchain_ollama import ChatOllama
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm = ChatOllama(model="llama3.2")
```

```python
llm.invoke("My name is  milan rai").content
```

```python
llm.invoke("What is my name?").content
```

```python
messages = ['My name is Nitish'] # this messages act as a state and also called conversation buffer
```

```python
output = llm.invoke(messages).content
output
```

```python
messages.append(output)
messages
```
