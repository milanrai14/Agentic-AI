```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class BatsmanState(TypedDict):
    runs: int 
    balls: int 
    fours: int 
    sixes: int

    sr: float
    bpb: float
    boundary_percent: float
    summary: str
```

```python
def calculate_sr(state: BatsmanState)-> BatsmanState: 
    sr = (state['runs']/state['balls']) /100

    state['sr'] = sr
    return sr
```

```python
def calculate_bpb(state: BatsmanState):
    bpb = state['balls']/(state['fours']+state['sixes'])
    state['bpb'] = bpb
    return state
```

```python
def calculate_boundary_percent(state: BatsmanState): 
    boundary_percent = (((state['fours']* 4+ state['sixes']*6)) / state['runs']) * 100
    state['boundary_boundary'] = boundary_percent
    return boundary_percent
```

```python
def summary(state: BatsmanState):
    summary = f"""
strike Rate = {state['sr']} \n 
Balls per boundart - {state['bpb']} \n 
boundary percent - {state['boundary_percent']}
    """
    state['summary'] = summary
```

```python
graph = StateGraph(BatsmanState)

graph.add_node("calculate_sr", calculate_sr)
graph.add_node("calculate_bpb", calculate_bpb)
graph.add_node("calculate_boundary_percent", calculate_boundary_percent)
graph.add_node("summary", summary)

#edges 
graph.add_edge(START, "calculate_sr")
graph.add_edge(START, "calculate_bpb")
graph.add_edge(START, "calculate_boundary_percent")

graph.add_edge('calcuate_sr', 'summary')
graph.add_edge('calcuate_bpb', 'summary')
graph.add_edge('calcuate_boundary_percent', 'summary')

graph.add_edge("summary", END)
```
